```python
import pandas as pd
import ast

# Dataset Load
df = pd.read_csv("disease_data.csv")

# String list ko actual list me convert karo
df['Symptom'] = df['Symptom'].apply(ast.literal_eval)

print("Dataset Loaded Successfully!")
print("Total Diseases:", len(df))
```

```python
import ipywidgets as widgets
from IPython.display import display
```

```python
# Symptoms Input Box

symptom_box = widgets.Textarea(
    value='',
    placeholder='Example:\nfever, cough, headache',
    description='Symptoms:',
    layout=widgets.Layout(width='700px', height='120px')
)

# Predict Button

predict_btn = widgets.Button(
    description='Predict Disease',
    button_style='success'
)

# Output Box

output = widgets.Output()
```

```python
def predict_disease(button):

    with output:

        output.clear_output()

        user_text = symptom_box.value.strip()

        if user_text == "":
            print("Please enter symptoms.")
            return

        user_symptoms = [
            symptom.strip().lower()
            for symptom in user_text.split(",")
        ]

        disease_scores = []

        for _, row in df.iterrows():

            disease = row["Disease"]

            disease_symptoms = [
                s.lower()
                for s in row["Symptom"]
            ]

            matched = len(
                set(user_symptoms) &
                set(disease_symptoms)
            )

            disease_scores.append(
                (disease, matched)
            )

        disease_scores.sort(
            key=lambda x: x[1],
            reverse=True
        )

        print("="*50)
        print("Predicted Disease")
        print("="*50)
        print(disease_scores[0][0])

        print("\nTop 5 Matches\n")

        for disease, score in disease_scores[:5]:

            print(
                f"{disease}  -->  {score} symptom(s) matched"
            )

predict_btn.on_click(predict_disease)
```

```python
display(symptom_box)
display(predict_btn)
display(output)
```
